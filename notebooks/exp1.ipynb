```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# models
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression

# metrics
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

# ml flow for model tracking
import mlflow
import mlflow.sklearn
```

```python
X = df.drop(columns=['Species', 'Id'])
y = LabelEncoder().fit_transform(df['Species'])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
import dagshub
mlflow.set_tracking_uri('https://dagshub.com/Lakshya0707/mlops-practice.mlflow')
dagshub.init(repo_owner= 'Lakshya0707', repo_name= 'mlops-practice', mlflow= True)

# mlflow experiment(logistic regression)
mlflow.set_experiment('Logistic Regression')
```

```python
# now scaling the data 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model_lor = LogisticRegression(max_iter=1000, random_state=42)
model_lor.fit(X_train_scaled, y_train)

y_pred = model_lor.predict(X_test_scaled)
print(classification_report(y_test, y_pred))
```
